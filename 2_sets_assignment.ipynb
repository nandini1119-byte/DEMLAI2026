```python

bank_branches = {"MG Road", "Indiranagar", "Whitefield", "Electronic City"}

customers = {"Rahul", "Anita", "Kiran", "Sneha", "Vikram"}

new_customers = {"Amit", "Priya", "Rahul", "Divya"}

loan_customers = {"Rahul", "Kiran", "Asha", "Ramesh"}

savings_customers = {"Rahul", "Sneha", "Priya", "Nikhil"}

transaction_ids = {101, 102, 103, 104, 105}

balances = {85000, 120000, 45000, 95000, 275000}

cities = {"Bangalore", "Hyderabad", "Chennai", "Pune"}

new_cities = {"Mumbai", "Bangalore", "Delhi"}

account_types = {"Savings", "Current", "Loan"}
```

# **Basic Operations**

```python
1#.Print the customers set.
print(customers)

2#.Find the number of elements in the customers set.
print(len(customers))

3#.Check whether "Rahul" exists in the customers set.
print("Rahul" in customers)

4#.Check whether "Meena" exists in the customers set.
print("Meena" in customers)

5#.Iterate through all elements in the bank_branches set.
for branch in bank_branches:
    print(branch)

```

```python
#6.Add "Meena" to the customers set.

customers.add("Meena")
print(customers)


#7.Add "Mysore" to the cities set.

cities.add("Mysore")
print(cities)
```

# **update()**

```python
#8.Update the customers set with new_customers.
customers.update(new_customers)
print(customers)

#9.Update the cities set with new_cities.
cities.update(new_cities)
print(cities)
```

```python
#10.Remove "Rahul" from the customers set.
customers.remove("Rahul")
print(customers)

#11.Remove "Pune" from the cities set.
cities.remove("Pune")
print(cities)
```

```python
#12.Discard "Anita" from the customers set.
customers.discard("Anita")
print(customers)

#13.Discard "Delhi" from the cities set.
cities.discard("Delhi")
print(cities)

#14.Discard "Kolkata" from the cities set.
cities.discard("Kolkata")
print(cities)
```

```python
#18.Create a copy of the customers set.
customers_copy = customers.copy()
print(customers_copy)

#19.Modify the copied set and verify that the original set remains unchanged.
customers_copy.add("Ramesh")
print(customers_copy)
print(customers)
```

# **union()**

```python
#20.Find the union of customers and new_customers.
union_customers = customers.union(new_customers)
print(union_customers)
```

# **intersection()**

```python
#21.Find the common customers between customers and new_customers.
common_customers = customers.intersection(new_customers)
print(common_customers)

#22.Find customers who have both Savings and Loan accounts.
loan_customers = {"Rahul", "Kiran", "Asha", "Ramesh"}
savings_customers = {"Rahul", "Sneha", "Priya", "Nikhil"}
both_accounts = loan_customers.intersection(savings_customers)
print(both_accounts)
```

# **difference()**

```python
#23.Find customers present only in customers but not in new_customers.
customers_only = customers.difference(new_customers)
print(customers_only)


#24.Find customers who have only Savings accounts.
savings_only = savings_customers.difference(loan_customers)
print(savings_only)
```

# **symmetric_difference()**

```python
#25.Find customers that are present in either customers or new_customers but not both.
symmetric_customers = customers.symmetric_difference(new_customers)
print(symmetric_customers)
```

# **issubset()**

```python
#26.Check whether {"Rahul", "Kiran"} is a subset of loan_customers.
subset_customers = {"Rahul", "Kiran"}.issubset(loan_customers)
print(subset_customers)
```

# **issuperset()**

```python
#27Check whether customers is a superset of {"Rahul", "Anita"}.
superset_customers = customers.issuperset({"Rahul", "Anita"})
print(superset_customers)
```

# **isdisjoint()**

```python
#28.Check whether loan_customers and new_cities are disjoint.
disjoint_sets = loan_customers.isdisjoint(new_cities)
print(disjoint_sets)
```
