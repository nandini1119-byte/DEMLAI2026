Find the highest profit from any single store-category combination and display the

## Question 40: Low-Margin Transactions

Management considers a profit margin below 22% to be weak. Find all store-category
combinations below that threshold and display the store, category, sales, cost, profit
and profit margin. Use Boolean indexing.

`np.where()` on a Boolean matrix returns two arrays - the row positions and the column
positions of every `True` - which are then paired to read the store and category names.

```python
profit_data = sales_data - cost_data

profit_margin = (profit_data / sales_data) * 100

weak_margin = profit_margin < 22

rows, cols = np.where(weak_margin)

for r, c in zip(rows, cols):
    print("Store       :", stores[r])
    print("Category    :", categories[c])
    print("Sales       :", sales_data[r, c])
    print("Cost        :", cost_data[r, c])
    print("Profit      :", profit_data[r, c])
    print("Margin (%)  :", round(profit_margin[r, c], 2))
    print("-" * 30)
```

## Question 45: Premium Performance

A combination is premium when sales are above 5,00,000 **and** profit is above
1,50,000. Find all premium-performing combinations.

The `&` operator combines two Boolean arrays element by element. Each condition needs
its own brackets because `&` binds tighter than `>`.

```python
profit_data = sales_data - cost_data

premium = (sales_data > 500000) & (profit_data > 150000)

rows, cols = np.where(premium)

for r, c in zip(rows, cols):
    print("Store    :", stores[r])
    print("Category :", categories[c])
    print("Sales    :", sales_data[r, c])
    print("Profit   :", profit_data[r, c])
    print("-" * 30)
```

## Question 47: Category Performance Classification

Classify every sales value as `"High"` (at least 5,00,000), `"Medium"` (2,50,000 to
4,99,999) or `"Low"` (below 2,50,000). Use `np.select()`.

`np.select()` checks the condition list in order and applies the first match, so the
conditions do not need to exclude one another manually.

```python
conditions = [
    sales_data >= 500000,
    (sales_data >= 250000) & (sales_data < 500000),

]

choices = ["High", "Medium"]

sales_level = np.select(conditions, choices, default="Unknown")

print(sales_level)
```
